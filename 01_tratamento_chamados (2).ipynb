```python
df_chamados = spark.table("workspace.default.chamados_suporte")

display(df_chamados)
```

```python
df_sem_duplicatas = df_chamados.dropDuplicates()
```

```python
print("Antes:", df_chamados.count())
print("Depois:", df_sem_duplicatas.count())
```

```python
display(df_sem_duplicatas)
```

```python
from pyspark.sql.functions import col, trim, upper, when
```

```python
df_padronizado = df_sem_duplicatas.withColumn(
    "categoria",
    upper(trim(col("categoria")))
)

display(df_padronizado)
```

```python
df_tratado = df_padronizado.withColumn(
    "categoria",
    when(
        col("categoria").isNull() | (col("categoria") == ""),
        "NAO INFORMADA"
    ).otherwise(col("categoria"))
)

display(df_tratado)
```

```python
df_tratado.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.silver_chamados")
```

```python
from pyspark.sql.functions import col

df_silver = spark.table("workspace.default.silver_chamados")
```

```python
df_tempo = df_silver.withColumn(
    "tempo_resolucao_horas",
    (
        col("data_fechamento").cast("long")
        - col("data_abertura").cast("long")
    ) / 3600
)
```

```python
display(
    df_tempo.select(
        "id_chamado",
        "data_abertura",
        "data_fechamento",
        "tempo_resolucao_horas"
    ).orderBy("id_chamado")
)
```

```python
df_sla = df_tempo.withColumn(
    "dentro_sla",
    col("tempo_resolucao_horas") <= col("sla_horas")
)
```

```python
display(
    df_sla.select(
        "id_chamado",
        "status",
        "tempo_resolucao_horas",
        "sla_horas",
        "dentro_sla"
    ).orderBy("id_chamado")
)
```

```python
df_sla.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.silver_chamados")
```

```python
%sql
SELECT
    categoria,
    COUNT(*) AS total_chamados
    FROM workspace.default.silver_chamados
    GROUP BY categoria
    ORDER BY total_chamados DESC;
```

```python
%sql
SELECT
    categoria,
    COUNT(*) AS chamados_resolvidos,
    ROUND(AVG(tempo_resolucao_horas), 2) AS media_horas
FROM workspace.default.silver_chamados
WHERE status = 'Resolvido'
  AND tempo_resolucao_horas IS NOT NULL
GROUP BY categoria
ORDER BY media_horas DESC;
```

```python
%sql
SELECT
    categoria,
    COUNT(*) AS chamados_avaliados,
    SUM(CASE WHEN dentro_sla = true THEN 1 ELSE 0 END) AS dentro_do_prazo,
    ROUND(
        100.0 * SUM(CASE WHEN dentro_sla = true THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS percentual_dentro_sla
FROM workspace.default.silver_chamados
WHERE status = 'Resolvido'
  AND dentro_sla IS NOT NULL
GROUP BY categoria
ORDER BY percentual_dentro_sla DESC;
```

```python
%sql
SELECT categoria, COUNT(*) AS chamados_pendentes
FROM workspace.default.silver_chamados
WHERE status IN ('Aberto','Em atendimento')
GROUP BY categoria
ORDER BY chamados_pendentes DESC;
```

```python
%sql
SELECT
    COUNT(*) AS total_registros,
    COUNT(DISTINCT id_chamado) AS ids_unicos,

    SUM(CASE
        WHEN id_chamado IS NULL THEN 1 ELSE 0
    END) AS ids_ausentes,

    SUM(CASE
        WHEN categoria IS NULL OR TRIM(categoria) = ''
        THEN 1 ELSE 0
    END) AS categorias_vazias,

    SUM(CASE
        WHEN tempo_resolucao_horas < 0 THEN 1 ELSE 0
    END) AS tempos_negativos,

    SUM(CASE
        WHEN status = 'Resolvido' AND data_fechamento IS NULL
        THEN 1 ELSE 0
    END) AS resolvidos_sem_fechamento

FROM workspace.default.silver_chamados;
```

```python
from pyspark.sql.functions import col, trim

df_validacao = spark.table("workspace.default.silver_chamados")

total = df_validacao.count()

assert total > 0, "A tabela Silver está vazia."

assert df_validacao.filter(
    col("id_chamado").isNull()
).count() == 0, "Existem chamados sem ID."

assert total == df_validacao.select(
    "id_chamado"
).distinct().count(), "Existem IDs duplicados."

assert df_validacao.filter(
    col("categoria").isNull() | (trim(col("categoria")) == "")
).count() == 0, "Existem categorias vazias."

assert df_validacao.filter(
    col("tempo_resolucao_horas") < 0
).count() == 0, "Existem tempos de resolução negativos."

assert df_validacao.filter(
    (col("status") == "Resolvido")
    & col("data_fechamento").isNull()
).count() == 0, "Existem chamados resolvidos sem fechamento."

print("Validações concluídas: dados aprovados nesses testes.")
```

```python
%sql
CREATE OR REPLACE TABLE workspace.default.gold_chamados_categoria
USING DELTA
AS
SELECT
    categoria,
    COUNT(*) AS total_chamados,

    SUM(CASE
        WHEN status = 'Resolvido' THEN 1 ELSE 0
    END) AS chamados_resolvidos,

    SUM(CASE
        WHEN status IN ('Aberto', 'Em atendimento') THEN 1 ELSE 0
    END) AS chamados_pendentes,

    ROUND(AVG(CASE
        WHEN status = 'Resolvido'
        THEN tempo_resolucao_horas
    END), 2) AS media_horas,

    SUM(CASE
        WHEN status = 'Resolvido' AND dentro_sla IS NOT NULL
        THEN 1 ELSE 0
    END) AS chamados_avaliados_sla,

    SUM(CASE
        WHEN status = 'Resolvido' AND dentro_sla = true
        THEN 1 ELSE 0
    END) AS chamados_dentro_sla,

    ROUND(100.0 * AVG(CASE
        WHEN status = 'Resolvido' AND dentro_sla = true THEN 1.0
        WHEN status = 'Resolvido' AND dentro_sla = false THEN 0.0
    END), 2) AS percentual_dentro_sla

FROM workspace.default.silver_chamados
GROUP BY categoria;
```
